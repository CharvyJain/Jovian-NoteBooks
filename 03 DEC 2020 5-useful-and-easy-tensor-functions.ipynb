## Function 1 - torch.transpose

**This function returns the transpose of the given tensor.**

```python
x = torch.tensor([[ 2.628, -0.98,  60.9],
        [5.1669,  2.99,  -0.4942]])

torch.transpose(x, 0, 1)
```

Here, We obtained a tensor where the rows and columns are interchanged.

```python
y = torch.tensor([[3.72,  12.225]])

torch.transpose(y, 0, 1)
```

We should give two parameters namely dim0 and dim1. These given dimensions will be swapped.

```python
x = torch.tensor([[ 2.628, -0.98,  60.9],
        [5.1669,  2.99,  -0.4942]])

torch.transpose(x, 0, 2)
```

We sholud not give the dimensions which are out of range. This will cause IndexError.

This function is very useful to deal with common problems in the field of linear algebra.

## Function 2 - torch.sort


**This funtcion sorts the given tensor along a given dimension. By default, it sorts in ascending order.**

```python
x = torch.tensor([[0.22,  0.0608,  6.719,  2.3332],
        [0.93,  -0.01,  0.68,  0.947],
        [0.5071,  0.43,  -0.553,  -1.0960]])

sorted, indices = torch.sort(x)
print(sorted)
```

As expected, this function returned a tensor with the values being sorted in ascending order.

```python
y = torch.tensor([[50.71, 0.22,  0.6719, -0.93],
        [ 0.608,  -0.0061,  -0.9497,  0.33],
        [ 0.58,  -0.9553,  1.0960,  -2.3332]])

sorted, indices = torch.sort(x, 1)
print(sorted)
print(indices)
```

The ourput tensor has values which are the sorted and indices are indices of the elements in the original input tensor. That's why there are jumbled.
We had specified the dimension to sort along.

```python
a = torch.tensor([98, 36, 56, 21, 39, 76, 36])
torch.sort(a, 1)
```

This error is shown because dimensional parameter has the value which does not exist for the tensor dimension. We have to be careful when giving the dimensional parameter.

This function is used to get a sorted data in ascending or descending order for our further calculations.

## Function 3 - torch.mean

**It returns the mean value of all elements in the input tensor.**

```python
x = torch.tensor([[ 0.2294, -0.5481,  1.3288]])

torch.mean(x)
```

Here we have got the mean value of the given tensor.

```python
y = torch.tensor([[ 4.19,  -0.743, -0.6921, -0.2577],
        [-1.8135,  3.8, -0.76,  0.1105],
        [ 0.01, -0.88, 7.07,  0.80]])

torch.mean(y, 1)
```

Here, We have given paramenter to reduce the dimension of output tensor.

```python
y = torch.tensor([[ 4.19,  -0.743, -0.6921, -0.2577],
        [-1.8135,  3.8, -0.76,  0.1105],
        [ 0.01, -0.88, 7.07,  0.80]])

torch.mean(y, 3 , keepdim=False) 
```

The range should be in between -2 to 1.

This function is used to calculate the mean of the given tensor.

## Function 4 - torch.chunk 

**This function splits a tensor into a specific number of chunks (splitting). Each chunk is a view of the input tensor.**

```python
x = torch.tensor([6.3, 69, 3.6, 56, 42, 89])

torch.chunk(x, 3, dim=0) 
```

We have given parameter 3, so the given tensor was splitted into 3 different smaller tensors.

```python
y = torch.tensor([300, 35, 563, 40, 5, 70, 75, 80, 80, 96])

torch.chunk(y, 3, 0)
```

Note that the third tensor has only two elements instead of three. This happens becaues the total values for the given tensor is odd number.

```python
r = torch.tensor([[2.0 , 5, 7, 5, 77, 2142, 11],[55, 67.96, 43]])

torch.chunk(r, 4, dim=0) 
```

Length of all the dim should be same.

This function can be used to manipulate the data of the tensors according to our needs.

## Function 5 - torch.det

**This function calculates determinant of a square matrix.**

```python
x = torch.tensor([[86, 3.6], [96.3, 89]])

torch.det(x)
```

The determinent of given tensor is 7307.3203

```python
y = torch.tensor([[[ 0.9254, -0.6213],
         [-0.5787,  1.6843]],

        [[ 0.3242, -0.9665],
         [ 0.4539, -0.0887]],

        [[ 1.1336, -0.4025],
         [-0.7089,  0.9032]]])
torch.det(y)
```

The determinant is calculated by calculating the determinant for each square matrix.

```python
x = torch.tensor([[1, 2], [1, 2, 3]])

torch.det(x)
```

In order to calculate the determinant, the matrix should be square.

This function is used for finding determinants of the square matrices.
